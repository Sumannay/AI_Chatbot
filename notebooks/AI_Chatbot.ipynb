```python
def chat_stream(question):
    chat_history.append({
        "role": "user",
        "content": question
    })

    stream = client.chat.completions.create(
        model=MODEL_NAME,
        messages=chat_history,
        stream=True
    )

    full_response = ""

    for chunk in stream:
        content = chunk.choices[0].delta.content

        if content:
            print(content, end="", flush=True)
            full_response += content

    print()

    chat_history.append({
        "role": "assistant",
        "content": full_response
    })

    return full_response
```

```python
def create_chunks(text, chunk_size=1000, overlap=200):
    chunks = []

    start = 0

    while start < len(text):
        end = start + chunk_size

        chunk = text[start:end].strip()

        if chunk:
            chunks.append(chunk)

        start += chunk_size - overlap

    return chunks
```

```python
from pathlib import Path
from pypdf import PdfReader

DOCUMENT_DIR = Path("../data/documents")


def load_documents(directory):
    documents = []

    pdf_files = list(directory.glob("*.pdf"))

    print(f"Found {len(pdf_files)} PDF file(s).")

    for pdf_path in pdf_files:

        print(f"Reading: {pdf_path.name}")

        reader = PdfReader(str(pdf_path))

        text = ""

        for page in reader.pages:

            page_text = page.extract_text()

            if page_text:
                text += page_text + "\n"

        if text.strip():

            documents.append({
                "source": pdf_path.name,
                "text": text
            })

    return documents
```

```python
def create_document_chunks(documents, chunk_size=1000, overlap=200):

    all_chunks = []

    for document in documents:

        text = document["text"]
        source = document["source"]

        start = 0

        while start < len(text):

            end = start + chunk_size

            chunk = text[start:end].strip()

            if chunk:

                all_chunks.append({
                    "text": chunk,
                    "source": source
                })

            start += chunk_size - overlap

    return all_chunks
```

```python
from pypdf import PdfReader

documents = []

for pdf_path in pdf_files:

    reader = PdfReader(str(pdf_path))

    text = ""

    for page in reader.pages:

        page_text = page.extract_text()

        if page_text:
            text += page_text + "\n"

    if text.strip():

        documents.append({
            "source": pdf_path.name,
            "text": text
        })

print("Documents loaded:", len(documents))

for document in documents:
    print(
        document["source"],
        "→",
        len(document["text"]),
        "characters"
    )
```

```python
def create_document_chunks(
    documents,
    chunk_size=1000,
    overlap=200
):

    all_chunks = []

    for document in documents:

        text = document["text"]
        source = document["source"]

        start = 0

        while start < len(text):

            end = start + chunk_size

            chunk = text[start:end].strip()

            if chunk:

                all_chunks.append({
                    "text": chunk,
                    "source": source
                })

            start += chunk_size - overlap

    return all_chunks
```

```python
document_chunks = create_document_chunks(documents)

print("Total chunks:", len(document_chunks))
```

```python
from sentence_transformers import SentenceTransformer

embedding_model = SentenceTransformer(
    "sentence-transformers/all-MiniLM-L6-v2"
)

print("Embedding model loaded successfully!")
```

```python
texts = [
    chunk["text"]
    for chunk in document_chunks
]

document_embeddings = embedding_model.encode(
    texts,
    convert_to_numpy=True
)

print("Embeddings created successfully!")
print("Embedding shape:", document_embeddings.shape)
```

```python
dimension = document_embeddings.shape[1]

document_index = faiss.IndexFlatL2(
    dimension
)

document_index.add(
    document_embeddings.astype("float32")
)

print("FAISS index created successfully!")
print("Number of vectors:", document_index.ntotal)
```

```python
INDEX_PATH = PROCESSED_DIR / "documents.index"

faiss.write_index(
    document_index,
    str(INDEX_PATH)
)

print("FAISS index saved!")
print(INDEX_PATH.resolve())
```

```python
CHUNKS_PATH = PROCESSED_DIR / "document_chunks.pkl"

with open(CHUNKS_PATH, "wb") as file:
    pickle.dump(
        document_chunks,
        file
    )

print("Document chunks saved!")
print(CHUNKS_PATH.resolve())
```

```python
print("===================================")
print("RAG DATABASE CREATED SUCCESSFULLY")
print("===================================")

print("FAISS index exists:",
      INDEX_PATH.exists())

print("Chunks file exists:",
      CHUNKS_PATH.exists())

print("Vectors:",
      document_index.ntotal)

print("Chunks:",
      len(document_chunks))
```

```python
import faiss
import pickle
from pathlib import Path

PROCESSED_DIR = Path("../data/processed")

INDEX_PATH = PROCESSED_DIR / "documents.index"
CHUNKS_PATH = PROCESSED_DIR / "document_chunks.pkl"

loaded_index = faiss.read_index(
    str(INDEX_PATH)
)

print("FAISS database loaded!")
print("Number of vectors:", loaded_index.ntotal)
```

```python
with open(CHUNKS_PATH, "rb") as file:
    loaded_chunks = pickle.load(file)

print("Document chunks loaded!")
print("Number of chunks:", len(loaded_chunks))
```

```python
print("First chunk:")
print("--------------------")

print("Source:", loaded_chunks[0]["source"])
print("Text:", loaded_chunks[0]["text"][:500])
```

```python
def retrieve_documents(question, k=3):

    question_embedding = embedding_model.encode(
        [question],
        convert_to_numpy=True
    ).astype("float32")

    distances, indices = loaded_index.search(
        question_embedding,
        k
    )

    results = []

    for distance, idx in zip(
        distances[0],
        indices[0]
    ):

        if idx == -1:
            continue

        results.append({
            "text": loaded_chunks[idx]["text"],
            "source": loaded_chunks[idx]["source"],
            "distance": float(distance)
        })

    return results
```

```python
results = retrieve_documents(
    "What programming languages does Suman know?"
)

for i, result in enumerate(results, start=1):

    print(f"\n===== RESULT {i} =====")
    print("Source:", result["source"])
    print("Distance:", result["distance"])
    print("Text:")
    print(result["text"][:500])
```

```python
chat_history = []

print("Chat memory initialized!")
```

```python
def chat_model(messages):

    response = client.chat.completions.create(
        model=MODEL_NAME,
        messages=messages,
        temperature=0.3
    )

    return response.choices[0].message.content
```

```python
SYSTEM_PROMPT = """
You are AI Chat, an intelligent and helpful AI assistant.

Your goals:
- Give accurate and useful answers.
- Explain difficult concepts clearly.
- Use simple language when appropriate.
- Provide examples when helpful.
- Be honest when uncertain.
- Do not invent facts.
- Follow the user's instructions.
"""

def chat(question):

    messages = [
        {
            "role": "system",
            "content": SYSTEM_PROMPT
        }
    ]

    messages.extend(chat_history)

    messages.append({
        "role": "user",
        "content": question
    })

    answer = chat_model(messages)

    chat_history.append({
        "role": "user",
        "content": question
    })

    chat_history.append({
        "role": "assistant",
        "content": answer
    })

    return answer
```

```python
answer = chat("My name is Suman.")

print("AI:", answer)
```

```python
def build_rag_prompt(question, results):

    context = "\n\n".join(
        f"Source: {result['source']}\n"
        f"{result['text']}"
        for result in results
    )

    return f"""
You are an intelligent AI assistant.

Use the document context below when it is relevant.

Rules:
- Answer using the provided context when the question is document-related.
- Do not invent information.
- If the document does not contain the answer, say so.
- For normal conversation, answer naturally.
- Keep responses clear and useful.

DOCUMENT CONTEXT:

{context}

USER QUESTION:

{question}
"""
```

```python
def chat_with_rag(question, k=3):

    results = retrieve_documents(
        question,
        k=k
    )

    rag_prompt = build_rag_prompt(
        question,
        results
    )

    messages = [
        {
            "role": "system",
            "content": SYSTEM_PROMPT
        }
    ]

    messages.extend(chat_history)

    messages.append({
        "role": "user",
        "content": rag_prompt
    })

    answer = chat_model(messages)

    chat_history.append({
        "role": "user",
        "content": question
    })

    chat_history.append({
        "role": "assistant",
        "content": answer
    })

    return {
        "answer": answer,
        "sources": [
            result["source"]
            for result in results
        ]
    }
```

```python
response = chat_with_rag(
    "What programming languages does Suman know?"
)

print("AI:")
print(response["answer"])

print("\nSources:")

for source in response["sources"]:
    print("-", source)
```
